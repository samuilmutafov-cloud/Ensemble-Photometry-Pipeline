```python
import os
import glob
import numpy as np
from astropy.io import fits
from astropy.nddata import CCDData
import ccdproc
from astroscrappy import detect_cosmics

# Базов път до папките на диск D:
base_dir = "D:/astronomy_pipeline/data/"
raw_dir = os.path.join(base_dir, "raw")
output_dir = os.path.join(base_dir, "reduced")

os.makedirs(output_dir, exist_ok=True)

def get_fits_files(folder_path):
    files = []
    for ext in ["*.fits", "*.fit", "*.FITS", "*.FIT"]:
        files.extend(glob.glob(os.path.join(folder_path, ext)))
    return files

print("🚀 Стартиране на конвейера за прецизна редукция на диск D:...\n")

# 1. MASTER BIAS
bias_files = get_fits_files(os.path.join(raw_dir, "bias"))
if bias_files:
    print(f"📦 Намерени {len(bias_files)} bias кадъра. Комбиниране...")
    bias_ccds = [CCDData.read(f, unit="adu") for f in bias_files]
    master_bias = ccdproc.combine(bias_ccds, method='median', sigma_clip=True, mem_limit=1e9)
    master_bias.write(os.path.join(output_dir, "master_bias.fits"), overwrite=True)
    print("✅ Master Bias е записан.")
else:
    print("⚠️ Липсват Bias кадри. Ще се използва стойност 0.")
    master_bias = None

# 2. MASTER DARK
dark_files = get_fits_files(os.path.join(raw_dir, "darks"))
if dark_files:
    print(f"📦 Намерени {len(dark_files)} dark кадъра. Комбиниране...")
    dark_ccds = []
    for f in dark_files:
        ccd = CCDData.read(f, unit="adu")
        if master_bias is not None:
            ccd.data = ccd.data - master_bias.data
        dark_ccds.append(ccd)
    master_dark = ccdproc.combine(dark_ccds, method='median', sigma_clip=True, mem_limit=1e9)
    master_dark.write(os.path.join(output_dir, "master_dark.fits"), overwrite=True)
    print("✅ Master Dark е записан.")
else:
    print("⚠️ Липсват Dark кадри. Прескачане.")
    master_dark = None

# 3. MASTER FLAT
flat_files = get_fits_files(os.path.join(raw_dir, "flats"))
if flat_files:
    print(f"📦 Намерени {len(flat_files)} flat кадъра. Комбиниране...")
    flat_ccds = []
    for f in flat_files:
        ccd = CCDData.read(f, unit="adu")
        if master_bias is not None:
            ccd.data = ccd.data - master_bias.data
        if master_dark is not None:
            ccd.data = ccd.data - master_dark.data
        flat_ccds.append(ccd)
    
    def scaling_func(arr): return 1.0 / np.ma.median(arr)
    master_flat = ccdproc.combine(flat_ccds, method='median', scale=scaling_func, sigma_clip=True, mem_limit=1e9)
    master_flat.write(os.path.join(output_dir, "master_flat.fits"), overwrite=True)
    print("✅ Master Flat е записан.")
else:
    print("⚠️ Липсват Flat кадри. Прескачане.")
    master_flat = None

# 4. ОБРАБОТКА НА НАУЧНИТЕ КАДРИ
science_files = get_fits_files(os.path.join(raw_dir, "science"))
if not science_files:
    print("❌ Няма научни кадри за обработка.")
else:
    print(f"\n🎬 Обработка на научните кадри и премахване на космически лъчи...")
    for f in science_files:
        fname = os.path.basename(f)
        sci_ccd = CCDData.read(f, unit="adu")
        
        if master_bias is not None:
            sci_ccd.data = sci_ccd.data - master_bias.data
        if master_dark is not None:
            sci_ccd.data = sci_ccd.data - master_dark.data
        if master_flat is not None:
            flat_data = master_flat.data.copy()
            flat_data[flat_data == 0] = 1.0
            sci_ccd.data = sci_ccd.data / flat_data
            
        gain = sci_ccd.header.get('GAIN', 1.0)
        rdnoise = sci_ccd.header.get('RDNOISE', 5.0)
        
        crmask, clean_data = detect_cosmics(
            sci_ccd.data, gain=gain, readnoise=rdnoise,
            sigclip=4.5, sigfrac=0.3, objlim=5.0, cleantype='meanmask'
        )
        sci_ccd.data = clean_data
        sci_ccd.header['CR_COUNT'] = (int(np.sum(crmask)), "Number of cosmic rays removed")
        
        out_path = os.path.join(output_dir, "proc_" + fname)
        sci_ccd.write(out_path, overwrite=True)
        print(f" ✨ {fname} -> РЕДУЦИРАН! (Космически лъчи: {np.sum(crmask)})")

print("\n🎉 СТЪПКА 1 Е НАПЪЛНО ЗАВЪРШЕНА!")
```

```python
import os
import glob
from astropy.io import fits
import astroalign as aa

reduced_dir = "D:/astronomy_pipeline/data/reduced/"
aligned_dir = "D:/astronomy_pipeline/data/aligned/"
os.makedirs(aligned_dir, exist_ok=True)

science_files = glob.glob(os.path.join(reduced_dir, "proc_*.fit*"))

if len(science_files) < 2:
    print(f"⚠️ Намерени са {len(science_files)} файла. За напасване са нужни поне 2 кадъра.")
    if len(science_files) == 1:
        fname = os.path.basename(science_files[0])
        with fits.open(science_files[0]) as hdul:
            hdul.writeto(os.path.join(aligned_dir, "aligned_" + fname), overwrite=True)
        print("Кадърът е записан директно в папка aligned.")
else:
    print(f"📦 Намерени са {len(science_files)} калибрирани кадъра за напасване.")
    
    ref_file = science_files[0]
    print(f"🎯 Референтен кадър (Шаблон): {os.path.basename(ref_file)}")
    
    with fits.open(ref_file) as ref_hdul:
        # КОРЕКЦИЯ: Добавяме [0] за достъп до данните в нулевия HDU блок
        ref_data = ref_hdul[0].data.astype('float64')
        ref_hdul.writeto(os.path.join(aligned_dir, "aligned_" + os.path.basename(ref_file)), overwrite=True)
    
    for target_file in science_files[1:]:
        fname = os.path.basename(target_file)
        print(f"🔄 Напасване на {fname} спрямо шаблона...")
        
        with fits.open(target_file) as tgt_hdul:
            # КОРЕКЦИЯ: Добавяме [0] за достъп до данните и хедъра
            tgt_data = tgt_hdul[0].data.astype('float64')
            tgt_header = tgt_hdul[0].header
            
            try:
                registered_data, footprint = aa.register(tgt_data, ref_data)
                hdu_new = fits.PrimaryHDU(data=registered_data.astype('float32'), header=tgt_header)
                hdu_new.header['ALIGNED'] = (True, "Image spatially aligned using astroalign")
                
                output_path = os.path.join(aligned_dir, "aligned_" + fname)
                hdu_new.writeto(output_path, overwrite=True)
                print(f"  ✨ {fname} е напаснат субпикселно и записан!")
            except Exception as e:
                print(f" ❌ Грешка при напасването на {fname}: {e}")

print("\n🎉 СТЪПКА 2 Е НАПЪЛНО ЗАВЪРШЕНА! Всички кадри в папка D:/astronomy_pipeline/data/aligned/ съвпадат пиксел по пиксел.")
```

```python
import os
import glob
from astropy.io import fits
from astropy.time import Time
from astropy.coordinates import SkyCoord, EarthLocation, AltAz
import astropy.units as u

# 1. ДЕФИНИРАНЕ НА КООРДИНАТИ НА ОБСЕРВАТОРИЯТА (Пример: Рожен, България)
# Ако наблюдавате от друго място, въведете вашите точни географски дължина, ширина и височина
telescope_location = EarthLocation(lon=24.7436 * u.deg, lat=41.6931 * u.deg, height=1750 * u.m)

# 2. Път до напаснатите кадри от Стъпка 2
aligned_dir = "D:/astronomy_pipeline/data/aligned/"
science_files = glob.glob(os.path.join(aligned_dir, "aligned_proc_*.fit*"))

print("🕰️ Стартиране на прецизната времева и координатна корекция...\n")

for f in science_files:
    fname = os.path.basename(f)
    
    with fits.open(f, mode='update') as hdul:
        header = hdul[0].header
        
        # Взимане на метаданните
        obj_name = header.get('OBJECT', 'Unknown')
        date_obs = header.get('DATE-OBS')
        exptime = header.get('EXPTIME', 0.0)
        
        # Изчисляване на средното време на експозицията (Mid-exposure UTC)
        t_start = Time(date_obs, format='isot', scale='utc')
        t_mid = t_start + (exptime / 2.0) * u.second
        
        # Взимане на Равноденствените координати (RA/DEC) директно от обекта
        # За максимална точност използваме SkyCoord
        ra = header.get('RA')
        dec = header.get('DEC')
        
        # Ако RA/DEC липсват като чисти градуси, скриптът ще потърси текстовия формат
        try:
            if isinstance(ra, str):
                sc = SkyCoord(ra, dec, unit=(u.hourangle, u.deg), frame='icrs')
            else:
                sc = SkyCoord(ra, dec, unit=(u.deg, u.deg), frame='icrs')
        except Exception:
            # Алтернатива: Автоматично намиране на координатите от вашия списък за V2494 Cyg
            sc = SkyCoord("20h58m21.1s", "+43d43m12s", frame='icrs') # Примерни координати за полето
            
        # 3. ИЗЧИСЛЯВАНЕ НА BJD_TDB
        ltt_bary = t_mid.light_travel_time(sc, location=telescope_location)
        t_bary = t_mid + ltt_bary
        bjd_tdb = t_bary.tdb.jd
        
        # 4. ИЗЧИСЛЯВАНЕ НА ВЪЗДУШНАТА МАСА (AIRMASS)
        altaz_frame = AltAz(obstime=t_mid, location=telescope_location)
        obj_altaz = sc.transform_to(altaz_frame)
        airmass = obj_altaz.secz.value # sec(z) приближение за въздушна маса
        
        # 5. ЗАПИСВАНЕ НА НОВИТЕ ДАННИ В FITS ХЕДЪРА
        header['BJD_TDB'] = (float(bjd_tdb), "Barycentric Julian Date (TDB scale) at mid-exposure")
        header['AIRMASS'] = (float(airmass), "Calculated Airmass at mid-exposure")
        
        # Информираме Astropy, че хедърът е променен и трябва да се запише на диска
        hdul.flush()
        
        print(f" ✨ {fname} ({obj_name}):")
        print(f"    -> Mid-Exposure UTC: {t_mid.isot}")
        print(f"    -> Научно BJD_TDB:   {bjd_tdb:.8f}")
        print(f"    -> Изчислена Airmass: {airmass:.4f}")

print("\n🎉 СТЪПКА 3 Е НАПЪЛНО ЗАВЪРШЕНА! Всички кадри вече имат релативистично време и въздушна маса в хедърите си.")
```

```python
import os
import glob
import numpy as np
import pandas as pd
from astropy.io import fits
from astropy.stats import sigma_clipped_stats
from photutils.detection import DAOStarFinder
from photutils.aperture import CircularAperture, CircularAnnulus, aperture_photometry

# 1. Път до напаснатите и времево коригирани кадри
aligned_dir = "D:/astronomy_pipeline/data/aligned/"
science_files = glob.glob(os.path.join(aligned_dir, "aligned_proc_*.fit*"))

# Папка за финалните резултати (текстови таблици)
output_data_dir = "D:/astronomy_pipeline/data/photometry_results/"
os.makedirs(output_data_dir, exist_ok=True)

print("🌟 Стартиране на автоматичното откриване на източници и фотометрия...\n")

for f in science_files:
    fname = os.path.basename(f)
    
    with fits.open(f) as hdul:
        # ДОБРА МЕДИЦИНА: Взимаме данните и хедъра от нулевия HDU блок
        data = hdul[0].data
        header = hdul[0].header
        
        bjd = header.get('BJD_TDB', 0.0)
        airmass = header.get('AIRMASS', 0.0)
        filt = header.get('FILTER', 'Unknown')
        obj_name = header.get('OBJECT', 'Unknown')
        
        # 2. Изчисляване на фоновия шум на небето (Sky Background & Noise)
        mean, median, std = sigma_clipped_stats(data, sigma=3.0)
        
        # 3. Търсене на звезди с DAOStarFinder
        finder = DAOStarFinder(fwhm=3.5, threshold=5.0*std)
        sources = finder(data - median)
        
        if sources is None or len(sources) == 0:
            print(f"❌ Не бяха открити звезди в кадъра {fname}")
            continue
            
        print(f"🎯 Открити са {len(sources)} звездни източника в кадър: {fname}")
        
        # 4. Взимане на координатите
        x_coords = sources['x_centroid']
        y_coords = sources['y_centroid']
        positions = np.transpose((x_coords, y_coords))
        
        # 5. Дефиниране на Динамична Апертура (Радиус = 5 пиксела)
        aperture_radius = 5.0
        apertures = CircularAperture(positions, r=aperture_radius)
        
        # Локален пръстен за измерване на небето около всяка звезда (Sky Annulus)
        annulus_apertures = CircularAnnulus(positions, r_in=10.0, r_out=15.0)
        
        # 6. Изпълнение на Фотометрията
        phot_table = aperture_photometry(data, apertures)
        annulus_phot = aperture_photometry(data, annulus_apertures)
        
        # Пресмятане на чистия блясък (изваждане на локалното небе)
        bkg_mean = annulus_phot['aperture_sum'] / annulus_apertures.area
        flux_cleaned = phot_table['aperture_sum'] - (bkg_mean * apertures.area)
        
        # 7. Изчисляване на Инструментални Звездни Величини (Instrumental Magnitudes)
        flux_cleaned[flux_cleaned <= 0] = 1.0
        instrumental_mag = -2.5 * np.log10(flux_cleaned)
        
        # Създаване на чиста финална таблица с Pandas
        df = pd.DataFrame({
            'Star_ID': sources['id'],
            'X_Pixel': x_coords,
            'Y_Pixel': y_coords,
            'Flux_Cleaned': flux_cleaned,
            'Mag_Inst': instrumental_mag,
            'Peak_Value': sources['peak'],
            'BJD_TDB': bjd,
            'Airmass': airmass,
            'Filter': filt
        })
        
        # Записваме резултатите в CSV файл за конкретния кадър
        csv_name = fname.replace('.fits', '.csv').replace('.fit', '.csv')
        csv_path = os.path.join(output_data_dir, "phot_" + csv_name)
        df.to_csv(csv_path, index=False)
        print(f"  💾 Резултатите са записани успешно в: phot_{csv_name}")

print("\n🎉 СТЪПКА 4 Е НАПЪЛНО ЗАВЪРШЕНА! Всички звездни източници са измерени фотометрично!")
```

```python
import pandas as pd
import glob
import os
import numpy as np

phot_dir = "D:/astronomy_pipeline/data/photometry_results/"
csv_files = glob.glob(os.path.join(phot_dir, "phot_aligned_proc_*.csv"))

print(f"📊 Анализ на {len(csv_files)} епохи...")

all_frames_data = []
for f in csv_files:
    df_frame = pd.read_csv(f)
    all_frames_data.append(df_frame)

df_all = pd.concat(all_frames_data, ignore_index=True)

# Автоматично подреждане и намиране на най-ярките и стабилни звезди (под прага от 65000 ADU за сигурност)
ref_frame_bjd = df_all['BJD_TDB'].min()
ref_frame = df_all[df_all['BJD_TDB'] == ref_frame_bjd]
good_stars = ref_frame[(ref_frame['Peak_Value'] > 10000) & (ref_frame['Peak_Value'] < 64000)]
brightest_stable = good_stars.sort_values(by='Peak_Value', ascending=False).head(8)

ensemble_ids = brightest_stable['Star_ID'].tolist()
print(f"⭐ Автоматично избран ансамбъл от стабилни сравнителни звезди (IDs): {ensemble_ids}")

# Усредняване на ансамбъла за всяка епоха
ensemble_mags = {}
for bjd, group in df_all.groupby('BJD_TDB'):
    comp_stars = group[group['Star_ID'].isin(ensemble_ids)]
    if not comp_stars.empty:
        ensemble_mags[bjd] = comp_stars['Mag_Inst'].mean()

# Изчисляване на Delta Mag
df_all['Delta_Mag'] = df_all.apply(lambda r: r['Mag_Inst'] - ensemble_mags[r['BJD_TDB']] if r['BJD_TDB'] in ensemble_mags else np.nan, axis=1)

output_path = os.path.join(phot_dir, "FINAL_LIGHTCURVE_DATA.csv")
final_df = df_all[['BJD_TDB', 'Filter', 'Airmass', 'Star_ID', 'X_Pixel', 'Y_Pixel', 'Mag_Inst', 'Delta_Mag', 'Peak_Value']].sort_values(by=['Star_ID', 'BJD_TDB'])
final_df.to_csv(output_path, index=False)

print("👑 АНСАМБЛОВАТА ФОТОМЕТРИЯ Е ОБНОВЕНА ЗА ВСИЧКИ КАДРИ!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

output_path = "D:/astronomy_pipeline/data/photometry_results/FINAL_LIGHTCURVE_DATA.csv"
df = pd.read_csv(output_path)

# Изчисляваме статистиката за променливост
star_stats = df.groupby('Star_ID').agg(
    X=('X_Pixel', 'first'),
    Y=('Y_Pixel', 'first'),
    Std_Dev=('Delta_Mag', 'std'),
    Mean_Mag=('Delta_Mag', 'mean'),
    Peak=('Peak_Value', 'max')
).dropna()

variable_candidates = star_stats[(star_stats['Peak'] > 5000) & (star_stats['Peak'] < 60000)].sort_values(by='Std_Dev', ascending=False)

if not variable_candidates.empty:
    target_id = variable_candidates.index[0]
    target_data = df[df['Star_ID'] == target_id].sort_values(by='BJD_TDB')

    # Изчисляваме релативно време в дни спрямо първия кадър за красива ос
    t0 = target_data['BJD_TDB'].min()
    relative_days = target_data['BJD_TDB'] - t0

    # ЧЕРТАНЕ НА НАУЧНАТА ГРАФИКА
    plt.figure(figsize=(10, 5))
    plt.plot(relative_days, target_data['Delta_Mag'], 'o-', color='crimson', markersize=8, linewidth=2, label=f'V2494 Cyg (Star ID: {target_id})')

    # Обръщане на астрономическата ос
    plt.gca().invert_yaxis()

    plt.title('Кратковременна оптическа променливост (V2494 Cyg Field)', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel(f'Време от началото на серията (Дни) [T0 = BJD {t0:.4f}]', fontsize=12)
    # Поправка на LaTeX синтаксиса чрез r'...'
    plt.ylabel(r'Диференциална звездна величина ($\Delta$ Mag)', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.legend(fontsize=11, loc='upper left')
    plt.tight_layout()

    # Записване с високо качество
    plt.savefig("D:/astronomy_pipeline/data/photometry_results/light_curve_fixed.png", dpi=300)
    plt.show()
    print(f"📈 Коригираната графика е записана! Базово време на първия кадър: BJD {t0:.6f}")
else:
    print("❌ Няма достатъчно данни за изчисляване на променливост.")
```
